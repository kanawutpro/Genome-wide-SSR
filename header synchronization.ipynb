```python
from Bio import SeqIO

def modify_header(records):
    for record in records:
        # Split the description by spaces
        description_parts = record.description.split()
        # Find the part that starts with 'scaffold'
        scaffold_part = next((part for part in description_parts if part.startswith('scaffold')), None)
        if scaffold_part:
            # Remove the comma at the end
            scaffold_part = scaffold_part.rstrip(',')
            # Change the id and description of the record
            record.id = scaffold_part
            record.description = scaffold_part
        yield record

# Read the original FASTA file
records = SeqIO.parse("Tocci_1.0_genomic.fasta", "fasta")

# Modify the headers
modified_records = modify_header(records)

# Write the modified records to a new FASTA file
with open("modified_output.fasta", "w") as output_handle:
    SeqIO.write(modified_records, output_handle, "fasta")

print("Done!")
```
